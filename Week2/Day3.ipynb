# Pandas Basics – Reading and Analyzing the Iris Dataset

## Aim

The aim of this notebook is to learn the basic operations of the Pandas library using the Iris dataset. The notebook demonstrates how to read data from a CSV file, display records, obtain dataset information, select specific columns, and filter rows based on user-defined conditions.

```python
import pandas as pd
```

# Reading the Dataset

The Iris dataset is loaded into a Pandas DataFrame using the `read_csv()` function. The **Id** column is used as the index of the DataFrame.

```python
data = pd.read_csv("Iris.csv", index_col="Id")

print(data)
```

### Explanation

- `pd.read_csv()` reads data from a CSV file.
- `index_col="Id"` specifies that the **Id** column should be used as the index instead of creating a default numerical index.
- The dataset is stored in the variable `data`.
# Displaying the First 10 Rows

The `head()` function is used to display the first few records of the dataset.

```python
print(data.head(10))
```

### Explanation

- `head(10)` displays the first **10 rows** of the dataset.
- This helps in quickly understanding the structure and contents of the data.
# Dataset Information

The `info()` function provides a summary of the dataset.

```python
print(data.info())
```

### Explanation

The `info()` function displays:

- Number of rows and columns
- Column names
- Data types of each column
- Number of non-null values
- Memory usage

This information helps identify missing values and understand the structure of the dataset.
# Selecting a Column

A single column can be selected by specifying its name inside square brackets.

```python
print(data["Species"])
```

### Explanation

- `data["Species"]` selects only the **Species** column.
- The output is displayed as a Pandas Series.
# Displaying the Entire Column

When a dataset is large, Pandas truncates the output. The `to_string()` function displays all the values without truncation.

```python
print(data["Species"].to_string())
```

### Explanation

- `to_string()` prints every value in the selected column.
- This is useful when you need to view the complete contents of a column.
# Filtering Rows

Rows can be filtered using logical conditions. In this example, the dataset is filtered based on the **Petal Length** values entered by the user.

```python
print("Filtering data according to Petal Length")

min_petal_length = float(input("Enter the minimum petal length: "))
max_petal_length = float(input("Enter the maximum petal length: "))

filtered_data = data[
    (data["PetalLengthCm"] > min_petal_length) &
    (data["PetalLengthCm"] <= max_petal_length)
]

print(filtered_data)
```

### Explanation

- `input()` accepts the minimum and maximum petal length from the user.
- `&` represents the logical **AND** operator.
- Only rows where:
  - **PetalLengthCm > Minimum Value**
  - **PetalLengthCm ≤ Maximum Value**

  are displayed.

This technique is commonly used for filtering datasets based on multiple conditions.
# Conclusion

The dataset was successfully loaded from a CSV file, the first few records were displayed, dataset information was examined, individual columns were selected, and rows were filtered based on user-defined conditions. These operations demonstrate the fundamental features of Pandas for data loading, exploration, and filtering, which are essential steps in data analysis and machine learning workflows.
