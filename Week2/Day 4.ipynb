# Pandas Data Cleaning using the Netflix Dataset

## Aim

The aim of this notebook is to perform basic data cleaning operations on the Netflix Titles dataset using the Pandas library. The notebook demonstrates how to detect missing values, fill null values, rename columns, remove duplicate records, and convert data types to prepare the dataset for analysis.

```python
import pandas as pd
```

# Loading the Dataset

The Netflix dataset is loaded from a CSV file into a Pandas DataFrame for data cleaning and preprocessing.

```python
Data = pd.read_csv("netflix_titles_duration_minutes.csv")
```

# Detecting Missing Values

Before cleaning the dataset, it is important to identify missing values in each column. The `isnull()` function checks for missing values, while `sum()` counts the total number of missing values in each column.

```python
print("Missing Values Before Cleaning:\n")

print(Data.isnull().sum())
```

### Explanation

- `isnull()` returns **True** for missing values and **False** otherwise.
- `sum()` counts the total number of missing values present in each column.
- Detecting missing values helps determine which columns require cleaning.

# Filling Missing Values

Missing values are replaced with appropriate default values to improve the quality of the dataset.

```python
Data['director'] = Data['director'].fillna("Director not known")

Data[['date_added','rating']] = Data[['date_added','rating']].fillna("N/A")

Data['duration'] = Data['duration'].fillna(0)
```

### Explanation

- Missing values in the **director** column are replaced with `"Director not known"`.
- Missing values in **date_added** and **rating** are replaced with `"N/A"`.
- Missing values in the **duration** column are replaced with **0**.
- Filling missing values prevents errors during analysis and visualization.

# Verifying Missing Values

After filling the missing values, the dataset is checked again to ensure that all null values have been handled.

```python
print("Missing Values After Cleaning:\n")

print(Data.isnull().sum())
```

### Explanation

Running `isnull().sum()` again confirms whether all missing values have been successfully replaced.

# Renaming a Column

The **title** column is renamed to **Title** to improve readability and maintain consistent naming conventions.

```python
Data = Data.rename(columns={"title":"Title"})
```

### Explanation

The `rename()` function changes the name of an existing column without affecting the data stored within it.

# Removing Duplicate Records

Duplicate records are removed based on the **Title** column.

```python
Data = Data.drop_duplicates(subset=['Title'])
```

### Explanation

- `drop_duplicates()` removes repeated rows.
- The `subset` parameter specifies that duplicate titles should be removed while keeping the first occurrence.
- Removing duplicates improves the accuracy of data analysis.

# Converting Data Types

The **duration** column is converted into the **float** data type.

```python
Data = Data.astype({'duration': float})
```

### Explanation

- `astype()` changes the data type of a column.
- Converting **duration** to **float** allows numerical calculations such as averages, sums, and statistical analysis.

# Displaying the Cleaned Dataset

The cleaned dataset can now be viewed to verify that all preprocessing steps have been applied successfully.

```python
print(Data.head())
```

# Conclusion

In this notebook, the Netflix dataset was successfully cleaned using the Pandas library. Missing values were detected and replaced with suitable default values, a column was renamed for better readability, duplicate records were removed, and the **duration** column was converted to a numeric data type. These preprocessing steps improve the quality and consistency of the dataset, making it suitable for further analysis and visualization.
