# NumPy Random Matrices and Matrix Operations

## Aim

The aim of this notebook is to learn how to generate random matrices using NumPy and Python's `random` module, calculate row-wise and column-wise sums and means, and perform matrix transposition. These operations are fundamental in data analysis, linear algebra, and machine learning.

```python
import numpy as np
import random as rnd
```

# Generating Random Matrices

Random matrices are useful for testing algorithms, simulations, and performing mathematical computations. In this section, random matrices are generated using both NumPy and Python's built-in `random` module.

```python
# Random Matrix using NumPy

random_matrix = np.random.randint(1, 100, size=(3,3))

# Random Matrix using Python's random module

random_matrix2 = np.array([
    [rnd.randint(0,10) for j in range(3)]
    for i in range(3)
])

print("Random Matrix using NumPy")
print(random_matrix)

print("\nRandom Matrix using random module")
print(random_matrix2)
```

### Explanation

- `np.random.randint()` generates random integers within a specified range.
- `size=(3,3)` creates a matrix with 3 rows and 3 columns.
- The second matrix is created using Python's `random.randint()` inside a nested list comprehension and then converted into a NumPy array.
# Row Sum and Column Sum

NumPy allows easy computation of row-wise and column-wise sums using the `sum()` function with the `axis` parameter.

```python
matrix_1 = np.random.randint(1,100,size=(4,4))

print("Matrix")
print(matrix_1)

column_sum = matrix_1.sum(axis=0)

row_sum = matrix_1.sum(axis=1)

print("\nColumn Sum")
print(column_sum)

print("\nRow Sum")
print(row_sum)
```

### Explanation

- `axis=0` calculates the sum of each column.
- `axis=1` calculates the sum of each row.
- NumPy performs these calculations efficiently without using loops.
# Row Mean and Column Mean

The average value of each row and column is calculated using the `mean()` function.

```python
mean_col = np.mean(matrix_1, axis=0)

mean_row = np.mean(matrix_1, axis=1, keepdims=True)

print("Mean Along Columns")
print(mean_col)

print("\nMean Along Rows")
print(mean_row)
```

### Explanation

- `axis=0` computes the mean of each column.
- `axis=1` computes the mean of each row.
- `keepdims=True` preserves the original two-dimensional structure by keeping the result as a column vector.
# Matrix Transpose

The transpose of a matrix is obtained by interchanging its rows and columns.

```python
original_matrix = np.random.randint(1,100,size=(4,3))

print("Original Matrix")
print(original_matrix)

transposed_matrix = original_matrix.transpose()

transposed_matrix2 = np.transpose(original_matrix)

print("\nTranspose using transpose()")
print(transposed_matrix)

print("\nTranspose using np.transpose()")
print(transposed_matrix2)
```

### Explanation

- `transpose()` swaps the rows and columns of a matrix.
- `np.transpose()` performs the same operation as the `transpose()` method.
- If the original matrix has dimensions **4 × 3**, the transposed matrix will have dimensions **3 × 4**.

Finally, matrix transposition was performed using two different methods.
